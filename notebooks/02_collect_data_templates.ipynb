## 1. Telegram Public Channel Collector

**Requirements to enable:**
1. `pip install telethon`
2. Create a Telegram app at https://my.telegram.org/apps
3. Add to `.env`:
   ```
   TELEGRAM_API_ID=your_id
   TELEGRAM_API_HASH=your_hash
   TELEGRAM_PHONE=+your_phone
   ```

> **Legal note:** Only collect from public channels. Do not collect private user data.

````python
import collect_telegram

# Check if credentials are available
has_creds = collect_telegram.check_credentials()
print(f"Telegram credentials available: {has_creds}")

if has_creds:
    print("To run full collection:")
    print("  from collect_telegram import collect_all")
    print("  collect_all(start_date='2023-09-01', end_date='2023-12-31')")
else:
    print("Running with sample data instead (see Notebook 01).")
````

## 3. Google Trends Collector

**Requirements:** `pip install pytrends`

> **Note:** pytrends is an **unofficial** wrapper for Google Trends. It may be rate-limited or blocked. Use with delays between requests.

````python
import collect_google_trends

has_pytrends = collect_google_trends.check_pytrends()
print(f"pytrends available: {has_pytrends}")

if has_pytrends:
    print("\nTo collect a single keyword:")
    print("  df = collect_google_trends.collect_keyword('hostages', '2023-09-01', '2023-12-31', geo='IL')")
else:
    print("Install with: pip install pytrends")
    print("Then run: python src/collect_google_trends.py")
````
