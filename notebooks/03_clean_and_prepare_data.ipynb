## 1. War Events

## 2. Telegram Data

```python
tg = clean_telegram(source='sample')

print(f"Shape: {tg.shape}")
print(f"Date range: {tg['date'].min()} to {tg['date'].max()}")
print(f"Channels: {tg['source_name'].unique()}")
print(f"\nNull counts:")
print(tg.isnull().sum())
```

```python
# Views distribution
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

tg['views'].hist(bins=50, ax=axes[0], color='#229ED9', alpha=0.75)
axes[0].set_title('Distribution of Message Views')
axes[0].set_xlabel('Views')

tg.groupby('source_name').size().sort_values().plot.barh(ax=axes[1], color='#229ED9', alpha=0.75)
axes[1].set_title('Message Count by Channel')
axes[1].set_xlabel('Messages')

plt.tight_layout()
plt.show()
```

## 3. RSS Data

```python
rss = clean_rss(source='sample')

print(f"Shape: {rss.shape}")
print(f"Date range: {rss['date'].min()} to {rss['date'].max()}")
print(f"\nCategory distribution:")
print(rss['category'].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

rss['category'].value_counts().plot.bar(ax=axes[0], color='#E63946', alpha=0.75)
axes[0].set_title('Articles by Category')
axes[0].tick_params(axis='x', rotation=30)

rss.groupby('source_name').size().sort_values().plot.barh(ax=axes[1], color='#E63946', alpha=0.75)
axes[1].set_title('Articles by Source')

plt.tight_layout()
plt.show()
```

## 4. Google Trends Data

```python
trends = clean_trends(source='sample')

print(f"Shape: {trends.shape}")
print(f"\nValue range per keyword:")
print(trends.groupby('keyword')['trend_value'].agg(['min', 'max', 'mean']).round(1))
```

## 5. Data Quality Summary

Confirm all datasets are clean and within the study period.

```python
summary = {
    'Telegram': {'rows': len(tg), 'nulls': tg.isnull().sum().sum(), 'date_range': f"{tg['date'].min()} – {tg['date'].max()}"},
    'RSS': {'rows': len(rss), 'nulls': rss.isnull().sum().sum(), 'date_range': f"{rss['date'].min()} – {rss['date'].max()}"},
    'Trends': {'rows': len(trends), 'nulls': trends.isnull().sum().sum(), 'date_range': f"{trends['date'].min()} – {trends['date'].max()}"},
}

pd.DataFrame(summary).T
```
