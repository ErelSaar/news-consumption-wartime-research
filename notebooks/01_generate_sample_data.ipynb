# Notebook 01 — Generate Sample Data

This notebook generates all simulated datasets for the research project.

**All data is fictional and designed to demonstrate a research workflow.**

Study period: **2023-09-01 to 2023-12-31**  
Key date: **2023-10-07 (October 7 Attack)**

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
import matplotlib.pyplot as plt
from sample_data_generator import (
    generate_war_events,
    generate_telegram_data,
    generate_rss_data,
    generate_trends_data,
    generate_unified_daily,
)
```

## 1. War Events

```python
events = generate_war_events()
events
```

## 2. Telegram Sample Data

```python
tg = generate_telegram_data()
print(f"Shape: {tg.shape}")
tg.head()
```

```python
tg['date'] = pd.to_datetime(tg['published_at']).dt.date
daily_msgs = tg.groupby('date').size()

fig, ax = plt.subplots(figsize=(13, 4))
ax.plot(pd.to_datetime(daily_msgs.index), daily_msgs.values, color='#229ED9', linewidth=1.5)
ax.axvline(pd.Timestamp('2023-10-07'), color='red', linewidth=2, linestyle='--', label='Oct 7')
ax.set_title('Simulated Telegram Messages Per Day')
ax.set_ylabel('Messages')
ax.legend()
plt.tight_layout()
plt.show()
```

## 3. RSS Sample Data

```python
rss = generate_rss_data()
print(f"Shape: {rss.shape}")
rss.head()
```

```python
rss['date'] = pd.to_datetime(rss['published_at']).dt.date
daily_articles = rss.groupby('date').size()

fig, ax = plt.subplots(figsize=(13, 4))
ax.bar(pd.to_datetime(daily_articles.index), daily_articles.values, color='#E63946', alpha=0.75, width=0.9)
ax.axvline(pd.Timestamp('2023-10-07'), color='black', linewidth=2, linestyle='--', label='Oct 7')
ax.set_title('Simulated RSS Articles Per Day')
ax.set_ylabel('Articles')
ax.legend()
plt.tight_layout()
plt.show()
```

## 4. Google Trends Sample Data

```python
trends = generate_trends_data()
print(f"Shape: {trends.shape}")
trends.head()
```

```python
fig, ax = plt.subplots(figsize=(13, 5))
for kw in trends['keyword'].unique():
    subset = trends[trends['keyword'] == kw].sort_values('date')
    ax.plot(pd.to_datetime(subset['date']), subset['trend_value'], label=kw, linewidth=1.5)

ax.axvline(pd.Timestamp('2023-10-07'), color='red', linewidth=2, linestyle='--', label='Oct 7')
ax.set_title('Simulated Google Trends by Keyword (Israel)')
ax.set_ylabel('Trend Value (0-100)')
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```
